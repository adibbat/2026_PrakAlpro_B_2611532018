## 1.1 Latar Belakang
Pemrograman merupakan salah satu bagian penting dalam bidang ilmu komputer yang digunakan untuk memberikan instruksi kepada komputer agar dapat melakukan suatu proses sesuai dengan kebutuhan pengguna. ## 2.1 Dasar Teori
Tipe data dan variabel merupakan konsep dasar dalam pemrograman yang digunakan untuk menyimpan dan mengelola data. Tipe data menunjukkan jenis data yang digunakan dalam suatu program, sedangkan variabel digunakan sebagai tempat untuk menyimpan nilai yang dapat digunakan dalam proses pemrograman. Pemahaman mengenai tipe data dan variabel diperlukan agar data dapat diproses sesuai dengan karakteristik dan kebutuhan program.
1. **Boolean**
merupakan jenis data yang memiliki dua nilai, yaitu True dan False. Boolean digunakan untuk menyatakan suatu kondisi yang hanya memiliki dua kemungkinan, seperti benar atau salah. Dalam pemrograman, boolean banyak digunakan dalam pemeriksaan kondisi dan proses pengambilan keputusan.
2. **Input**
merupakan proses memasukkan atau menerima data dari pengguna ke dalam program. Data yang diberikan melalui input dapat disimpan ke dalam variabel dan selanjutnya digunakan dalam proses pengolahan data. Input memungkinkan program menerima informasi secara langsung dari pengguna.
3. **Konstanta**
merupakan nilai yang ditetapkan dalam program dan digunakan sebagai nilai tetap. Berbeda dengan variabel yang nilainya dapat berubah, konstanta digunakan ketika suatu nilai perlu dipertahankan agar tetap sama selama program digunakan. Konstanta dapat membantu menjaga konsistensi nilai tertentu dalam suatu program.
4. **Numerik**
merupakan data yang berupa angka dan digunakan untuk melakukan berbagai operasi perhitungan. Data numerik dapat digunakan dalam operasi matematika seperti penjumlahan, pengurangan, perkalian, dan pembagian. Data numerik menjadi salah satu jenis data penting dalam program yang melakukan pengolahan angka.
5. **String**
merupakan data yang digunakan untuk menyimpan teks atau kumpulan karakter. String dapat berupa huruf, kata, kalimat, angka yang diperlakukan sebagai teks, maupun karakter tertentu. Dalam pemrograman, string umumnya digunakan untuk menyimpan data berbentuk teks yang perlu ditampilkan atau diolah oleh program.

## 2.2 Kode Program

### 2.2.1 Program "Boolean_2611532018"

```python
# Buat file dengan nama Boolean_NIM.py
# nama variabel ditambah 4 digit nim terakhir contoh: nilai_1234
# Deklarasi variabel dengan tipe data Boolean
is_lulus_2018 = True
is_cumlaude_2018 = True

# Menggunakan Boolean
nilai_2018 = 85
batas_lulus_2018 = 75

# Menentukan nilai Boolean dari kondisi
status_kelulusan_2018 = nilai_2018 >= batas_lulus_2018 # Hasilnya akan True

print("=== Check Kelulusan ===")
print("Nilai:", nilai_2018)
print("Apakah Lulus?:", status_kelulusan_2018)
if is_lulus_2018 and is_cumlaude_2018:
    print("Selamat, Anda Lulus dengan Predikat Cum Laude!")
```

**Penjelasan :**
1. Baris 4, **is_lulus_2018 = True** berfungsi membuat variabel is_lulus_2018 dengan nilai True.
2. Baris 5, **is_cumlaude_2018 = True** berfungsi membuat variabel is_cumlaude_2018 dengan nilai True.
3. Baris 8, **nilai_2018 = 85** berfungsi menyimpan nilai mahasiswa.
4. Baris 9, **batas_lulus_2018 = 75** berfungsi menentukan nilai minimum untuk lulus.
5. Baris 12, **status_kelulusan_2018 = nilai_2018 >= batas_lulus_2018** berfungsi membandingkan nilai dengan batas lulus.
6. Baris 14, **print("=== Check Kelulusan ===")** berfungsi menampilkan tulisan Check Kelulusan sebagai judul.
7. Baris 15, **print("Nilai:", nilai_2018)** berfungsi menampilkan nilai mahasiswa, yaitu 85.
8. Baris 16, **print("Apakah Lulus?:", status_kelulusan_2018)** berfungsi menampilkan hasil pengecekan kelulusan.
9. Baris 17, **if is_lulus_2018 and is_cumlaude_2018:** Mengecek apakah kedua kondisi bernilai True.
and berarti kedua kondisi harus benar.
10. Baris 18, **print("Selamat, Anda Lulus dengan Predikat Cum Laude!")** Jika kedua kondisi benar, program menampilkan ucapan selamat, anda lulus dengan predikat Cum Laude.

### 2.2.2 Program "Input_2611532018"

```python
# Buat file dengan nama Input_NIM.py
# Buat program untuk menghitung luas persegi panjang
# Nama variabel ditambah 4 digit nim terakhir contoh: panjang_1234
# Program ini menggunakan fungsi input() untuk meminta pengguna
# memasukkan nilai panjang dan lebar persegi panjang.
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer
panjang_2018 = input('Masukan nilai panjang: ')
lebar_2018 = input('Masukkan nilai lebar: ')
luas_2018 = int(panjang_2018) * int(lebar_2018)
print("Luas:", luas_2018)
```

**Penjelasan :**
1. Baris 7, **panjang_2018 = input('Masukan nilai panjang: ')** berfungsi untuk meminta pengguna untuk memasukkan nilai panjang. Nilai tersebut disimpan dalam variabel panjang_2018
2. Baris 8: **lebar_2018 = input('Masukkan nilai lebar: ')** berfungsi meminta pengguna untuk memasukkan nilai lebar. Nilai tersebut disimpan dalam variabel lebar_2018.
3. Baris 9, **luas_2018 = int(panjang_2018) * int(lebar_2018)** berfungsi mengubah nilai panjang dan lebar menjadi integer menggunakan int(). Kemudian mengalikan panjang dengan lebar untuk mendapatkan luas persegi panjang. Rumusnya: Luas = panjang × lebar.
4. Baris 10, **print("Luas:", luas_2018)** berfungsi menampilkan hasil perhitungan luas ke layar.

### 2.2.3 Program "Konstanta_2611532018"

```python
# Buat file dengan nama Konstanta_NIM.py
# Program ini menggunakan konstanta untuk menghitung luas lingkaran
# nama variabel ditambah 4 digit nim terakhir contoh: jari_1234

from typing import Final
PI: Final = 3.14
print("pi: %f" % (PI))
jari_2018 = float(input('masukkan nilai jari-jari: '))
luas_2018 = PI * jari_2018 * jari_2018
print("Luas lingkaran dengan jari-jari %.2f adalah %.2f" % (jari_2018, luas_2018))
```

**Penjelasan :**
1. Baris 5, **from typing import Final** berfungsi untuk mengambil Final dari Python. Final digunakan untuk menandai bahwa suatu nilai dianggap sebagai konstanta dan tidak seharusnya diubah.
2. Baris 6, **PI: Final = 3.14** berfungsi untuk membuat konstanta PI dengan nilai 3.14.
Nilai ini digunakan dalam rumus luas lingkaran.
3. Baris 7, **print("pi: %f" % (PI))** berfungsi untuk menampilkan nilai PI ke layar. %f digunakan untuk menampilkan nilai dalam bentuk angka desimal.
4. Baris 8, **jari_2018 = float(input('masukkan nilai jari-jari: '))** berfungsi untuk meminta pengguna memasukkan nilai jari-jari lingkaran. input() digunakan untuk menerima masukan dari pengguna. float() digunakan agar nilai yang dimasukkan dapat berupa angka desimal. Nilai tersebut disimpan dalam variabel jari_2018.
5. Baris 9, **luas_2018 = PI * jari_2018 * jari_2018** berfungsi untuk menghitung luas lingkaran menggunakan rumus: Luas = π × r × r. Hasil perhitungan disimpan dalam variabel luas_2018.
6. Baris 10, **print("Luas lingkaran dengan jari-jari %.2f adalah %.2f" % (jari_2018, luas_2018))** berfungsi untuk menampilkan hasil perhitungan luas lingkaran. %.2f digunakan untuk menampilkan angka dengan 2 angka di belakang koma.

### 2.2.4 Program "Numerik_2611532018"

```python
# Buat file dengan nama Numerik_NIM.py
# nama variabel ditambah 4 digit terakhir NIM contoh: a_5500
#create a variabel with numeric type value.
a_2018=100
print("The type of variable having value", a_2018, "is ", type(a_2018))

#create a variable with float value.
b_2018=10.2345
print("The type of variable having value", b_2018, "is ", type(b_2018))

#create a variable with complex value.
c_2018=100+3j
print("The type of variable having value", c_2018, "is ", type(c_2018))
```

**Penjelasan :**
1. Baris 4 dan 5, **a_2018=100** membuat variabel a_2018 dengan nilai 100, sedangkan **print("The type of variable having value", a_2018, "is ", type(a_2018))** menampilkan nilai dari a_2018 dan mengecek tipe datanya.
3. Baris 8 dan 9, **b_2018=10.2345** membuat variabel b_2018 dengan nilai 10.234, sedangkan **print("The type of variable having value", b_2018, "is ", type(b_2018))** menampilkan nilai b_2018 dan mengecek tipe datanya.
5. Baris 12 dan 13, **c_2018=100+3j** membuat variabel c_2018 dengan nilai 100 + 3j, sedangkan **print("The type of variable having value", c_2018, "is ", type(c_2018))** menampilkan nilai c_2018 dan mengecek tipe datanya.
