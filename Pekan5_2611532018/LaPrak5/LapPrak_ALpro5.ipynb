## 2.1 Dasar Teori
Perulangan (looping) adalah struktur kontrol pemrograman yang digunakan untuk mengeksekusi blok kode secara berulang-ulang selama kondisi tertentu terpenuhi atau sampai rentang iterasi selesai.

1. Perulangan for dan Fungsi range():
    Mengontrol jalannya iterasi berdasarkan rentang angka yang ditentukan. Fungsi range() menentukan batas awal, akhir, dan langkah iterasi, sedangkan parameter end pada print() mengatur agar output dicetak menyamping tanpa berpindah baris secara otomatis.   

2. Perulangan Bersarang (Nested Loop):
    Struktur di mana perulangan berada di dalam perulangan lain. Outer loop (perulangan luar) bertindak sebagai pengontrol baris, sedangkan inner loop (perulangan dalam) bertindak sebagai pengontrol kolom atau karakter dalam satu baris.   

3. Operasi Akumulasi dan Kondisional:
    Penggunaan variabel penampung di dalam iterasi untuk menghitung jumlah total (misal jumlah = jumlah + i), serta penggunaan percabangan if-else dan operator modulus % untuk menyaring kondisi tertentu seperti bilangan genap atau ganjil.   

4. Pemetaan Pola Matematis (Loop Tables):
    Metode menentukan persamaan matematika yang menghubungkan variabel baris (outer loop) dengan jumlah karakter (inner loop) untuk menghasilkan pola atau bentuk tertentu.

### 2.2.3 Program "perulangan_for3_2611532018"

```python
# Buat  file dengan nama perulangan_for3_nim.py
# Buat program untuk perulangan for dalam python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

ulang_2018 = int(input("Masukkan jumlah perulangan: "))

jumlah_2018 = 0
for i_2018 in range(1, ulang_2018 + 1):
    print(i_2018, end=" ")
    jumlah_2018 = jumlah_2018 + i_2018

    if i_2018 < ulang_2018:
        print(" + ", end=" ")
    else:
        print(" = ", jumlah_2018,end=" ")
print()
print("Jumlah =", jumlah_2018)
```

3. Baris 9 berfungsi untuk memulai perulangan *for* yang berjalan mulai dari angka `1` hingga `ulang_2018`.
4. Baris 10 berfungsi untuk mencetak nilai variabel `i_2018` secara menyamping dengan pemisah spasi menggunakan `end=" "`.
5. Baris 11 berfungsi untuk menambahkan nilai `i_2018` saat ini ke dalam variabel `jumlah_2018` pada setiap iterasi.
6. Baris 13 berfungsi untuk mengecek apakah nilai `i_2018` masih lebih kecil dari `ulang_2018` (bukan iterasi terakhir).
7. Baris 14 berfungsi untuk mencetak tanda pemisah `" + "` secara menyamping jika kondisi pada baris 13 terpenuhi.
8. Baris 15 berfungsi untuk menentukan blok alternatif jika iterasi telah mencapai angka terakhir (`i_2018` sama dengan `ulang_2018`).
9. Baris 16 berfungsi untuk mencetak tanda `" = "` beserta total hasil akumulasi pada `jumlah_2018` secara menyamping.

### 2.2.5 Program "nested_for1_2611532018"

```python
# Buat  file dengan nama nested_for1_nim.py
# Buat program untuk perulangan for dalam python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

batas_2018 = int(input("Masukkan nilai batas: "))
for line_2018 in range(1, batas_2018 + 1):
    for j_2018 in range(1, (-1 * line_2018 + batas_2018) + 1):
        print(".", end="")
    print(line_2018)  
```

**Penjelasan**

3. Baris 8 berfungsi untuk memulai *inner loop* (perulangan dalam) yang menghitung dan mencetak jumlah titik pada setiap baris berdasarkan rumus `(-1 * line_2018 + batas_2018)`.
4. Baris 9 berfungsi untuk mencetak karakter titik `"."` secara menyamping tanpa berpindah baris menggunakan `end=""`.

### 2.2.8 Program "nested_for4_2611532018"

```python
# Buat  file dengan nama nested_for4_nim.py
# Buat program untuk perulangan for dalam python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

tinggi_2018 = int(input("Masukkan tinggi pola (bilangan genap, misal 10): "))

if tinggi_2018 % 2 != 0:
    print("Tinggi harus bilangan genap!")
else:
    a_2018 = tinggi_2018
    c_2018 = a_2018
    lebar_2018 = (2 * tinggi_2018) - 2

    for i_2018 in range(1, tinggi_2018 + 1):
        b_2018 = c_2018 + 1

        for j_2018 in range(1, lebar_2018 + 1):

            #Baris atas dan bawah
            if i_2018 == 1 or i_2018 == tinggi_2018:
                if j_2018 == 1 or j_2018 == lebar_2018:
                    print("#", end="")
                else:
                    print("=", end="")

            #Baris isi
            else:
                if j_2018 == 1 or j_2018 == lebar_2018:
                    print("|", end="")
                else:
                    if j_2018 == c_2018:
                        print("<", end="")
                    elif j_2018 == b_2018:
                        print(">", end="")
                    elif j_2018 == (lebar_2018 - c_2018):
                        print("<", end="")
                    elif j_2018 == (lebar_2018 - c_2018 + 1):
                        print(">", end="")
                    elif j_2018 > b_2018 and j_2018 < (lebar_2018 - c_2018):
                        print(".", end="")
                    else:
                        print(" ", end="")

        print()

        #Logika asli java
        a_2018 -= 2

        if a_2018 <= 0:
            c_2018 = (-a_2018) + 2
        else:
            c_2018 = a_2018
```

**Penjelasan**

Berdasarkan praktikum yang telah dilakukan, dapat disimpulkan bahwa struktur perulangan *for* pada Python merupakan mekanisme kontrol yang sangat efektif untuk mengeksekusi sekumpulan perintah secara berulang-ulang berdasarkan rentang angka atau urutan data yang ditentukan secara presisi menggunakan fungsi `range()`. Kombinasi antara perulangan, operasi kondisional `if-else`, operator pembanding/modulus, serta logika akumulasi variabel memungkinkan pemrosesan data numerik maupun pembuatan deret hitung dilakukan secara efisien. Selain itu, penerapan struktur perulangan bersarang (*nested loop*) terbukti dapat menyelesaikan permasalahan pemetaan pola geometris dan matriks dua dimensi melalui pembagian peran antara perulangan luar (*outer loop*) sebagai pengontrol baris dan perulangan dalam (*inner loop*) sebagai pengontrol kolom.

Dalam konteks pemrograman Python, konsep ini memanfaatkan koordinasi antara indeks iterasi dan persamaan matematika (seperti pada *loop table*) sehingga pembentukan output yang kompleks, penyaringan data, dan penghitungan akumulasi total dapat dijalankan secara otomatis, sistematis, dan terstruktur tanpa perlu menuliskan kode pemrosesan secara berulang.
