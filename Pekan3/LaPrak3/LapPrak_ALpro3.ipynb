## 2.1 Dasar Teori
Operator pada Python adalah simbol yang digunakan untuk mengeksekusi operasi tertentu terhadap operan (variabel dan nilai). Berikut adalah kategori operator pada Python:

1. Operator Aritmatika: Digunakan untuk operasi matematika. Terdiri dari Penjumlahan (+), Pengurangan (-), Perkalian (*), Pembagian (/), Pembagian bulat (//), Modulus atau sisa bagi (%), dan Pangkat (**).
2. Operator Perbandingan (Comparison): Digunakan untuk membandingkan dua buah nilai dan akan menghasilkan nilai boolean True atau False (==, !=, >, <, >=, <=).
3. Operator Logika: Digunakan untuk menggabungkan beberapa kondisi (and, or, not).
4. Operator Penugasan (Assignment): Digunakan untuk memberikan nilai awal atau memperbarui nilai suatu variabel (=, +=, -=, *=, /=, %=, **=, //=).
5. Operator Keanggotaan (Membership): Menggunakan operator `in` (ada di dalam koleksi) dan `not in` (tidak ada di dalam koleksi) untuk memeriksa posisi suatu elemen di dalam sekumpulan data.
6. Operator Identitas (Identity): Digunakan untuk membandingkan lokasi objek langsung di dalam memori menggunakan `is` (objek yang sama) dan `is not` (objek berbeda).
7. Operator Bitwise: Operator yang mengeksekusi data pada level bit (biner) yang diwakili oleh lambang AND (&), OR (|), XOR (^), NOT (~), Geser kiri (<<), dan Geser kanan (>>).

### 2.2.4 Program "assignment_2611532018"

```python
# Buat file dengan nama assignment_NIM.py
# Nama variabel ditambahkan 4 digit nim terakhir contoh: angka1234
# Program ini menggunakan fungsi input()
# Nilai yang dimasukkan akan dikonversikan menjadi tipe data integer
# Program operator assignment dalam Python

angka1_2018 = int(input("Input angka-1: ")) 
angka2_2018 = int(input("Input angka-2: "))

print("\nNilai awal angka1 =", angka1_2018)
print("Nilai angka2 =", angka2_2018)

# Assignment biasa
hasil_2018 = angka1_2018
print("\nAssignment biasa (=)")
print("Hasil =", hasil_2018)

# Assignment penambahan
hasil_2018 = angka1_2018
hasil_2018 += angka2_2018
print("\nAssignment penambahan (+=)")
print("Hasil =")

# Assignment pengurangan
hasil_2018 = angka1_2018
hasil_2018 -= angka2_2018
print("\nAssignment pengurangan (-=)")
print("Hasil =", hasil_2018)

# Assignment perkalian
hasil_2018 = angka1_2018
hasil_2018 *= angka2_2018
print("\nAssignment perkalian (*=)")
print("Hasil =", hasil_2018)

# Assignment pembagian, pembagian bulat, dan sisa bagi
if angka2_2018 != 0:
    hasil_2018 = angka1_2018
    hasil_2018 /= angka2_2018
    print("\nAssignment pembagian (/=)")
    print("Hasil =", hasil_2018)
    # Operator tambahan
    hasil_2018 = angka1_2018
    hasil_2018 //= angka2_2018
    print("\nAssignment pembagian bulat (//=)")
    print("Hasil =", hasil_2018)
    hasil_2018 = angka1_2018
    hasil_2018 %= angka2_2018
    print("\nAssignment sisa bagi (%=)")
    print("Hasil =", hasil_2018)
else:
    print("\nPembagian tidak dapat dilakukan")
    print("Angka kedua tidak boleh bernilai 0")

# Operator tambahan: assignment perpangkatan
hasil_2018 = angka1_2018
hasil_2018 **= angka2_2018
print("\nAssignment perpangkatan (**=)")
print("Hasil =", hasil_2018)
```

### 2.2.6 Program "bitwise_2611532018"

```python
# Buat file dengan nama bitwise_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menggunakan fungsi input()

print("\n========================================")
print("3. OPERATOR BITWISE")
print("========================================")

angka1_2018 = int(input("Masukkan angka bitwise-1: "))
angka2_2018 = int(input("Masukkan angka bitwise-2: "))

print("\nAngka dalam bentuk desimal dan biner")
print("angka1 =", angka1_2018, "| biner =", bin(angka1_2018))
print("angka2 =", angka2_2018, "| biner =", bin(angka2_2018))

# Bitwise AND
hasil_2018 = angka1_2018 & angka2_2018
print("\nBitwise AND (&)")
print(angka1_2018, "&", angka2_2018, "=", hasil_2018)
print("Biner hasil =", bin(hasil_2018))
print("Biner hasil (8 bit) =", format(hasil_2018, "08b"))

# Bitwise OR
hasil_2018 = angka1_2018 | angka2_2018
print("\nBitwise OR (|)")
print(angka1_2018, "|", angka2_2018, "=", hasil_2018)
print("Biner hasil =", bin(hasil_2018))
print("Biner hasil (8 bit) =", format(hasil_2018, "08b"))

# Bitwise XOR
hasil_2018 = angka1_2018 ^ angka2_2018
print("\nBitwise XOR (^)")
print(angka1_2018, "^", angka2_2018, "=", hasil_2018)
print("Biner hasil =", bin(hasil_2018))
print("Biner hasil (8 bit) =", format(hasil_2018, "08b"))

# Bitwise NOT
hasil_2018 = ~angka1_2018
print("\nBitwise NOT (~)")
print("~", angka1_2018, "=", hasil_2018)
print("Biner hasil =", bin(hasil_2018))
print("Biner hasil (8 bit) =", format(hasil_2018, "08b"))

# Bitwise geser kiri
jumlah_geser_2018 = int(input("\nMasukkan jumlah pergeseran bit: "))

hasil_2018 = angka1_2018 << jumlah_geser_2018
print("\nBitwise geser kiri (<<)")
print(angka1_2018, "<<", jumlah_geser_2018, "=", hasil_2018)
print("Biner hasil =", bin(hasil_2018))
print("Biner hasil (8 bit) =", format(hasil_2018, "08b"))

# Bitwise geser kanan
hasil_2018 = angka1_2018 >> jumlah_geser_2018
print("\nBitwise geser kanan (>>)")
print(angka1_2018, ">>", jumlah_geser_2018, "=", hasil_2018)
print("Biner hasil =", bin(hasil_2018))
print("Biner hasil (8 bit) =", format(hasil_2018, "08b"))
```
