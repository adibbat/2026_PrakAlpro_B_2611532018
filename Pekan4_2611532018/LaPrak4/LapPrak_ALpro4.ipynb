## 2.1 Dasar Teori
Struktur percabangan (control flow) dalam Python adalah blok kode yang memungkinkan sebuah program untuk mengevaluasi satu atau lebih kondisi (ekspresi boolean) dan mengambil keputusan untuk mengeksekusi instruksi tertentu berdasarkan hasil evaluasi tersebut (True atau False). Terdapat beberapa jenis struktur percabangan yang umum digunakan sesuai dengan kompleksitas logika masalah

1. Percabangan Tunggal (if)
Struktur ini merupakan bentuk percabangan yang paling sederhana. Blok instruksi di dalam pernyataan if hanya akan dieksekusi apabila kondisi yang dievaluasi bernilai benar (True). Jika kondisi bernilai salah (False), program akan mengabaikan blok kode tersebut dan langsung melanjutkan ke baris instruksi berikutnya di luar percabangan.
2. Percabangan Ganda (if-else)
Struktur if-else menyediakan dua jalur eksekusi yang saling berlawanan. Jika kondisi pada blok if bernilai True, maka instruksi di dalam blok tersebut akan dijalankan. Namun, jika kondisi tersebut bernilai False, program akan secara otomatis beralih dan menjalankan blok instruksi yang berada di dalam pernyataan else sebagai alternatif keluaran.
3. Percabangan Majemuk (if-elif-else)
Struktur ini digunakan untuk menangani lebih dari dua kondisi yang saling berkaitan atau eksklusif satu sama lain (mutually exclusive). Program akan mengevaluasi setiap kondisi secara berurutan dari atas ke bawah. Begitu menemukan satu kondisi yang bernilai True, blok instruksi terkait akan dieksekusi dan program akan langsung keluar dari struktur percabangan (kondisi di bawahnya tidak akan dicek lagi). Blok else biasanya ditambahkan di akhir sebagai penangkap kondisi default.
4. Percabangan Independen (Multi-IF)
Berbeda dengan percabangan majemuk, struktur Multi-IF merupakan susunan beberapa pernyataan if yang berdiri sendiri tanpa adanya ikatan logika berantai (tanpa elif atau else). Dalam struktur ini, program akan secara mutlak mengevaluasi setiap blok if satu per satu. Struktur ini sangat cocok diimplementasikan pada kasus yang bersifat akumulatif, seperti penambahan diskon bertumpuk atau sistem skor yang variabel syaratnya bisa terpenuhi lebih dari satu secara bersamaan.
5. Pencocokan Pola (match-case)
Diperkenalkan sejak Python versi 3.10, match-case merupakan mekanisme struktural yang menyerupai switch-case pada bahasa pemrograman C atau Java. match-case bekerja dengan cara mencocokkan nilai dari sebuah variabel (subjek) terhadap sekumpulan pola nilai atau case tertentu. Struktur ini membuat penulisan kode menjadi lebih bersih dan terstruktur ketika harus mencocokkan nilai spesifik secara presisi, dibandingkan jika harus menulis rentetan elif yang panjang. Kondisi sisa atau default ditangani menggunakan simbol underscore (_).

## 2.2 Kode Program

### 2.2.1 Program "if1_2611532018"

```python
# Buat  file dengan nama if1_nim.py
# Buat program untuk kondisional if
# Nama variabel ditambah 4 digit nim terakhir contoh: ipk_1234
# Program ini menggunakan fungsi input()

ipk_2018 = float(input("Inputkan IPK Anda = "))

if ipk_2018 > 2.75:
    print("Anda Lulus Sangat Memuaskan dengan IPK " + str(ipk_2018))
else:
    print("Anda Tidak Lulus")
print("Program Selesai")
```

**Penjelasan :**
1. Baris 6 kode ini, yaitu **ipk_2018 = float(input("Inputkan IPK Anda = "))** merupakan variabel yang akan menyimpan IPK yang diinputkan oleh user.
2. Baris 8 kode ini, yaitu **if ipk_2018 > 2.75:** merupakan struktur data percabangan, if(jika) akan memeriksa angka yang diinputkan user yang tersimpan pada variabel ``ipk_2018`` apakah nilainya lebih dari nilai yang ditetapkan yaitu 2.75.
3. Baris 9 kode ini akan menampilkan **Anda Lulus Sangat Memuaskan dengan IPK (angka yang diinputkan user)**, output ini akan ditampilkan bila baris 8, yaitu **if ipk_2018 > 2.75:** adalah ``TRUE``.
4. baris 10 dan 11, else(selain itu) akan bekerja dan menampilkan outpu **Anda Tidak Lulus** apabila **if ipk_2018 > 2.75:** ``FALSE``.
5. Baris 12 kode ini akan menampilkan **Program Selesai** yang menyatakan bahwa program telah terselesaikan.

### 2.2.2 Program "multi_if1_2611532018"

```python
# Buat  file dengan nama multi_if1_nim.py
# Buat program untuk kondisional if
# Nama variabel ditambah 4 digit nim terakhir contoh: ipk_1234
# Program ini menggunakan fungsi input()

umur_2018 = int(input("Inputkan Umur Anda = "))
sim_2018 = input("Apakah Anda sudah punya SIM C (y/t) = ")[0]

if umur_2018 >= 17 and sim_2018 == "y":
    print("Anda sudah dewasa dan boleh bawa motor")

if umur_2018 >= 17 and sim_2018 != "y":
    print("Anda sudah dewasa tetapi belum boleh bawa motor")

if umur_2018 < 17 and sim_2018 == "y":
    print("Anda belum Cukup Umur punya SIM")

if umur_2018 < 17 and sim_2018 != "y":
    print("Anda belum Cukup Umur bawa motor")
print("Program Selesai")
```

**Penjelasan :**
1. Baris 6 kode ini akan meminta user untuk menginputkan usia/umurnya lalu menyimpannya pada variabel ``umur_2018``.
2. Baris 7 kode ini akan meminta user untuk memilih anatara **y/t** (y=ya, t=tidak) dan menginputkan salah satunya lalu menyimpannya pada variabel ``sim_2018``.
3. Baris 9 hingga 19 akan mengambil data dari variabel ``umur_2018`` dan ``sim_2018`` lalu memeriksa kondisi if satu persatu apakah kondisi if True atau False dan menampilkan output yang kondisinya True, output yang dihasilkan tidak hanya satu, program akan menampilkan output dari setiap kondisi if yang True.
4. Baris 20 kode ini akan menampilkan **Program Selesai** yang menyatakan bahwa program telah terselesaikan.

### 2.2.3 Program "if_elif_else1_2611532018"

```python
# Buat  file dengan nama if_elif_else1_nim.py
# Buat program untuk kondisional if
# Nama variabel ditambah 4 digit nim terakhir contoh: ipk_1234
# Program ini menggunakan fungsi input()

umur_2018 = int(input("Inputkan Umur Anda = "))
sim_2018 = input("Apakah Anda sudah punya SIM C (y/t) = ")[0]

if umur_2018 >= 17 and sim_2018 == "y":
    print("Anda sudah dewasa dan boleh bawa motor")
elif umur_2018 >= 17 and sim_2018 != "y":
    print("Anda sudah dewasa tetapi tidak boleh bawa motor")
elif umur_2018 < 17 and sim_2018 == "y":
    print("Anda belum Cukup Umur punya SIM")
else:
    print("Anda belum Cukup Umur dan tidak boleh bawa motor")
print("Program Selesai")
```

**Penjelasan :**
1. Baris Baris 6 kode ini akan meminta user untuk menginputkan usia/umurnya lalu menyimpannya pada variabel ``umur_2018``.
2. Baris 7 kode ini akan meminta user untuk memilih anatara **y/t** (y=ya, t=tidak) dan menginputkan salah satunya lalu menyimpannya pada variabel ``sim_2018``.
3. Baris 9 hingga 16 akan mengambil data dari variabel ``umur_2018`` dan ``sim_2018`` lalu memeriksa kondisi dari atas apakah True atau False, pemeriksaan akan terhenti jika ada kondisi yang True dan tidak akan melanjutkan pemeriksaan lalu menampilkan output.
4. Baris 17 kode ini akan menampilkan **Program Selesai** yang menyatakan bahwa program telah terselesaikan.

### 2.2.4 Program "nilai_2611532018"

```python
# Buat  file dengan nama nilai_nim.py
# Buat program untuk nested if
# Nama variabel ditambah 4 digit nim terakhir contoh: ipk_1234
# Program ini menggunakan fungsi input()
# Program konversi nilai angka menjadi huruf

nilai_2018 = int(input("inputkan nilai angka= "))

if nilai_2018 >= 81:
    print("A")
elif nilai_2018 >= 70:
    print("B")
elif nilai_2018 >= 60:
    print("C")
elif nilai_2018 >= 50:
    print("D")
else:
    print("E")
```

**Penjelasan :**

2. Baris 9 hingga 18 akan mengambil data yang telah diinputkan user yang tersimpan pada variabel ``nilai_2018`` lalu memeriksa kondisi dari atas apakah True atau False, pemeriksaan akan terhenti jika ada kondisi yang True dan tidak akan melanjutkan pemeriksaan lalu menampilkan output.

### 2.2.5 Program "multi_if2_2611532018"

```python
# Buat  file dengan nama multi_if2_nim.py
# Buat program untuk kondisional if
# Nama variabel ditambah 4 digit nim terakhir contoh: ipk_1234
# Program ini menggunakan fungsi input()
# Program Menghitung Diskon Belanja

# Input dari user
total_belanja_2018 = float(input("Masukkan total belanja (Rp): "))

# input status member (mengecek apakah user mengetik 'y' atau 'ya')
input_member_2018 = input("Apakah Anda Member? (y/t): ").strip().lower()
is_member_2018 = input_member_2018 in ["y", "ya"]

# input status kode promo (mengecek apakah user mengetik 'y' atau 'ya')
input_promo_2018 = input("Apakah kode promo valid? (y/t): ").strip().lower()
is_promo_2018 = input_promo_2018 in ["y", "ya"]

total_diskon_persen_2018 = 0

# Multi-IF terpisah: Setiap kondisi diperiksa secara independen
# Diskon bisa ditumbuk (akumulasi) jika memenuhi beberapa syarat sekaligus

if total_belanja_2018 > 1000000:
    total_diskon_persen_2018 += 10 # Diskon belanja besar

if is_member_2018:
    total_diskon_persen_2018 += 5 # Diskon member

if is_promo_2018:
    total_diskon_persen_2018 += 15 # Diskon voucher

# Menghitung nominal diskon dan total bayar
nominal_diskon_2018 = total_belanja_2018 * (total_diskon_persen_2018 / 100)
total_bayar_2018 = total_belanja_2018 - nominal_diskon_2018

# Output hasil
print("\n--- Rincian Pembayaran ---")
print(f"Total Diskon  : {total_diskon_persen_2018}% (Rp {nominal_diskon_2018:,.0f})")
print(f"Total Bayar    : Rp {total_bayar_2018:,.0f}")

print(f"Total diskon yang anda dapatkan: {total_diskon_persen_2018}%")
# Output: Total diskon yang Anda dapatkan: 30% jika belanja > 1 juta, member, dan kode promo valid
```

**Penjelasan**
1. Baris 8 kode ini akan menampilkan pesan untuk meminta input total belanja, mengubah teks input tersebut menjadi tipe angka desimal (float), lalu menyimpannya ke variabel ``total_belanja_2018``.
2. Baris 11 dan 12 kode ini akan meminta input status member dari pengguna ("y/t"), membersihkan spasi dan mengubahnya ke huruf kecil, lalu mengevaluasinya menjadi nilai kebenaran (True/False) untuk disimpan pada variabel ``is_member_2018``.
3. Baris 15 dan 16 kode ini akan meminta input status kode promo dari pengguna, memproses teksnya, dan mengevaluasinya menjadi nilai True/False untuk disimpan pada variabel ``is_promo_2018``.
4. Baris 18 kode ini akan membuat variabel ``total_diskon_persen_2018`` dan memberikannya nilai awal 0, yang berfungsi sebagai tempat menampung akumulasi persentase diskon.
5. Baris 23 dan 24 kode ini akan memeriksa apakah total belanja lebih dari 1.000.000; jika ya, maka program akan menambahkan nilai 10 ke dalam total persentase diskon.
6. Baris 26 dan 27 kode ini akan memeriksa apakah pengguna adalah member (``is_member_2018`` bernilai True); jika ya, maka program akan menambahkan nilai 5 ke dalam total persentase diskon.
7. Baris 29 dan 30 kode ini akan memeriksa apakah kode promo valid (``is_promo_2018`` bernilai True); jika ya, maka program akan menambahkan nilai 15 ke dalam total persentase diskon.
8. Baris 33 dan 34 kode ini akan menghitung nominal potongan harga dalam Rupiah berdasarkan persentase diskon yang terkumpul, kemudian menghitung total akhir yang harus dibayar.
9. Baris 37 hingga 41 kode ini akan mencetak seluruh rincian pembayaran ke layar secara berurutan, meliputi persentase diskon, nominal potongan harga, dan total harga akhir yang harus dibayar oleh pengguna.

### 2.2.6 Program "match_case1_2611532018"

```python
# Buat  file dengan nama match_case1_nim.py
# Buat program untuk match case
# Nama variabel ditambah 4 digit nim terakhir contoh: ipk_1234
# Program ini menggunakan fungsi input()
# Program konversi angka menjadi nama bulan

bulan_2018 = int(input("Masukkan angka bulan (1-12): "))

match bulan_2018:
    case 1:
        print("Januari")
    case 2:
        print("Februari")
    case 3:
        print("Maret")
    case 4:
        print("April")
    case 5:
        print("Mei")
    case 6:
        print("Juni")
    case 7:
        print("Juli")
    case 8:
        print("Agustus")
    case 9:
        print("September")
    case 10:
        print("Oktober")
    case 11:
        print("November")
    case 12:
        print("Desember")
    case _:
        print("Angka tidak valid")
```

**Penjelasan :**
1. Baris 7 kode ini akan pengguna memasukkan sebuah nilai(1-12), kemudian menyimpannya ke dalam variabel ``bulan_2018``.
2. Baris 9 kode ini akan memulai struktur percabangan match untuk mengevaluasi nilai yang tersimpan di dalam variabel ``bulan_2018``.
3. Baris 10 hingga 33 kode ini akan memeriksa kecocokan nilai ``bulan_2018`` secara berurutan dari angka 1 hingga 12. Jika nilai tersebut cocok dengan salah satu kondisi (case 1 hingga case 12), maka program akan menampilkan nama bulan yang sesuai ke layar (mulai dari "Januari" untuk angka 1, hingga "Desember" untuk angka 12).
4. Baris 34 kode ini akan menangani kondisi default atau kondisi sisa (case _), yaitu bertindak sebagai penangkap apabila nilai dari ``bulan_2018`` tidak ada yang cocok dengan angka 1 hingga 12 yang telah didefinisikan sebelumnya.
5. Baris 35 kode ini akan mencetak teks "Angka tidak valid" ke layar jika blok case _ (kondisi pada baris 34) dieksekusi.
